```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import find_peaks as findPeaks
```

```python
hourly_counts = np.array([5,3,3,6,10,2,2,4,5,8,4,3,3,6,16,11,8,2,2,0,0,0,1,1])
print(hourly_counts)
print(hourly_counts.shape)
```

```python
plt.plot(hourly_counts)
plt.xlabel('Time')
plt.ylabel('Amplitude')
```

```python
(Pks, _) = findPeaks(hourly_counts, prominence=10)
print(Pks)
```
